```python
import json
from collections import defaultdict

import pandas as pd

from src import FIGURES_DIR, RESULTS_DIR
```

```python
surrogate_dir = RESULTS_DIR / "surrogate_experimental"
json_files = list(surrogate_dir.glob("*.json"))
data_dict = defaultdict(list)

key_list = [
    "dim",
    "tol",
    "n_train",
    "n_test",
    "num_clusters",
    "error_metric",
    "error_eval_mode",
    "train_time_sec",
    "pred_time_sec",
    "num_initial_clusters",
]
for json_file in json_files:
    try:
        if json_file.stat().st_size == 0:
            print(f"Skipping empty JSON file: {json_file.name}")
            continue

        with open(json_file, "r", encoding="utf-8") as f:
            data = json.load(f)
    except (json.JSONDecodeError, OSError) as e:
        print(f"Skipping invalid JSON file: {json_file.name} ({e})")
        continue

    for key in key_list:
        data_dict[key].append(data[key])
    data_dict["e"].append(data["error_metrics"]["all"]["l2_rel_error"])
    data_dict["e_in"].append(data["error_metrics"]["in"]["l2_rel_error"])
    data_dict["p_out"].append(data["error_metrics"]["perct_out"])
    data_dict["N"].append(data["total_f_evals"])


df = pd.DataFrame(data_dict)
df = df.sort_values(
    by=["dim", "tol", "error_eval_mode", "error_metric"],
    ascending=[True, False, True, True],
)
rename = {
    "n_train": "N_0",
    "n_test": "N_test",
    "num_clusters": "N_c",
}
order = [
    "dim",
    "tol",
    "error_eval_mode",
    "error_metric",
    "N_0",
    "N_test",
    "N_c",
    "N",
    "e",
    "e_in",
    "p_out",
    "train_time_sec",
    "pred_time_sec",
    "num_initial_clusters",
]
df = df.rename(columns=rename)
df = df[order]
# df = df[df["dim"] == 5]
df.reset_index(drop=True, inplace=True)
display(df)
df.to_clipboard(index=False)
temp_df = df[
    ["dim", "tol", "N_0", "N_c", "error_eval_mode", "error_metric", "train_time_sec"]
]
# temp_df.to_clipboard(index=False)
# temp_df
```
